```python
import datetime
import ephem
from typing import List

def years_with_full_moon_on_date(target_month: int, target_day: int,
                                 start_year: int, end_year: int) -> List[int]:
    matching_years = []
    
    for year in range(start_year, end_year + 1):
        try:
            target_date = datetime.date(year, target_month, target_day)
        except ValueError:
            continue
            
        ephem_date = ephem.Date(target_date)
        prev_full = ephem.previous_full_moon(ephem_date)
        next_full = ephem.next_full_moon(ephem_date)
        
        # Convert Ephem dates to Python datetime objects
        # This handles the fractional day properly
        prev_dt = prev_full.datetime()
        next_dt = next_full.datetime()
        target_dt = target_date  # Already a datetime.date
        
        # Compare only the date parts (year, month, day)
        if (target_dt.year == prev_dt.year and 
            target_dt.month == prev_dt.month and 
            target_dt.day == prev_dt.day):
            matching_years.append(year)
        elif (target_dt.year == next_dt.year and 
              target_dt.month == next_dt.month and 
              target_dt.day == next_dt.day):
            matching_years.append(year)
    
    return matching_years

if __name__ == "__main__":
    # Test for August 31
    years = years_with_full_moon_on_date(8, 31, 1990, 2060)
    print(f"Years with full moon on August 31: {years}")
```

```python
import datetime
import ephem
import pandas as pd
from typing import List

def years_with_full_moon_on_date(target_month: int, target_day: int,
                                 start_year: int, end_year: int) -> List[int]:
    matching_years = []
    
    for year in range(start_year, end_year + 1):
        try:
            target_date = datetime.date(year, target_month, target_day)
        except ValueError:
            continue
            
        ephem_date = ephem.Date(target_date)
        prev_full = ephem.previous_full_moon(ephem_date)
        next_full = ephem.next_full_moon(ephem_date)
        
        prev_dt = prev_full.datetime()
        next_dt = next_full.datetime()
        
        if (target_date.year == prev_dt.year and 
            target_date.month == prev_dt.month and 
            target_date.day == prev_dt.day):
            matching_years.append(year)
        elif (target_date.year == next_dt.year and 
              target_date.month == next_dt.month and 
              target_date.day == next_dt.day):
            matching_years.append(year)
    
    return matching_years

# Create lists to store data
years = []
months = []
days = []
fm_years_list = []

# Generate all valid dates
start_year = 1900
end_year = 2200

for month in range(1, 13):
    for day in range(1, 32):
        # Skip invalid dates (like Feb 31, Apr 31, etc.)
        try:
            datetime.date(2000, month, day)  # Using 2000 as a test year
        except ValueError:
            continue
        
        # Calculate full moon years for this date
        fm_years = years_with_full_moon_on_date(month, day, start_year, end_year)
        
        # Store the data
        months.append(month)
        days.append(day)
        fm_years_list.append(fm_years)
        
        print(f"Processed {month}/{day}: {len(fm_years)} years")

# Create DataFrame
df = pd.DataFrame({
    'month': months,
    'day': days,
    'full_moon_years': fm_years_list
})

# Save to CSV
df.to_csv('moon_data.csv', index=False)
print(f"\nDone! Created moon_data.csv with {len(df)} rows")

# Show sample data
print("\nSample data:")
print(df.head())
```
